# 🧠 Notebook 01: 3D VisReg JEPA Self-Supervised Pre-training
### Part of the BraTS 3D Volumetric Multimodal JEPA Research Suite

This notebook runs **SSL pre-training only** (the long pole — run once):
1. **Self-Supervised Pre-training (100 Epochs, AMP)**: Single-encoder JEPA with decoupled Center, Scale (unit variance penalty), and 1D Sliced-Wasserstein Shape Regularization against standard normal quantiles.

> **Note:** SSL pretraining is label-free — v1 and v2 images are identical. Prefer running this on the v2 attachment so the checkpoint fingerprint matches v2 finetunes.
